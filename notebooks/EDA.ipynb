# AQI — Exploratory Data Analysis
Heatmaps, histograms, boxplots, pairplots, PCA scree plot.

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from preprocessing import load_dataset, clean_dataframe, cap_outliers_iqr
df = cap_outliers_iqr(clean_dataframe(load_dataset()))
df.describe()
```

```python
plt.figure(figsize=(9,7)); sns.heatmap(df.corr(), annot=True, cmap='coolwarm'); plt.title('Correlation Heatmap');
```

```python
df.hist(figsize=(12,8), bins=30); plt.tight_layout();
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from utils import FEATURE_COLUMNS
X = StandardScaler().fit_transform(df[FEATURE_COLUMNS])
pca = PCA().fit(X)
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o'); plt.axhline(0.95, ls='--'); plt.title('Cumulative Explained Variance'); plt.xlabel('Components'); plt.ylabel('Variance');
```
